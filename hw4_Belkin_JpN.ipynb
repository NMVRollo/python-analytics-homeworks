## Задание 1. Поиск владельца документа

Пользователь по команде `p` может узнать владельца документа по его номеру.

Пользовательский ввод обрабатывается в цикле `while` до ввода команды `q`.

````python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}
````

````python
def get_owner(documents, document_number):
    for document in documents:
        if document['number'] == document_number:
            return document['name']
    return None


def print_owner(documents):
    document_number = input('Введите номер документа: ')
    owner = get_owner(documents, document_number)

    if owner:
        print(f'Владелец документа: {owner}')
    else:
        print('Документ не найден в базе')


def main(documents):
    while True:
        command = input('Введите команду: ')

        if command == 'p':
            print_owner(documents)
        elif command == 'q':
            break
        else:
            print('Неизвестная команда')


main(documents)
````
