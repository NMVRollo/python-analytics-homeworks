## Задание 1. Анализ возраста покупателей

Загружаем первые 15 строк из `web_clients_correct.csv`. Для столбца `age` применяем функцию через `.apply()`: если возраст больше 25 лет — `Старше 25`, иначе — `Младше или равен 25`.

```python
import pandas as pd

df = pd.read_csv('web_clients_correct.csv', nrows=15)
df
```

```python
def age_group(age):
    if age > 25:
        return 'Старше 25'
    return 'Младше или равен 25'

df['age'] = df['age'].apply(age_group)
```

```python
print(df['age'].to_string())
```

## Задание 2. Фильтрация по полу и возрасту

Повторно загружаем исходные данные, чтобы столбец `age` снова содержал числовые значения. Отбираем покупателей мужского пола (`male`) старше 30 лет.

```python
df = pd.read_csv('web_clients_correct.csv', nrows=15)

result = df[(df['sex'] == 'male') & (df['age'] > 30)]
result
```
