## Задание. Формирование файла funnel.csv

Для пользователей из `visit_log.csv` определяется категория покупки по данным из `purchase_log.txt`. Файл `visit_log.csv` обрабатывается построчно и целиком в память не загружается. В `funnel.csv` записываются только визиты пользователей, у которых была покупка.

```python
import csv
import json

purchase_log = {}

with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    for line in f:
        purchase = json.loads(line)
        if purchase['user_id'] != 'user_id':
            purchase_log[purchase['user_id']] = purchase['category']
```

Создаём `funnel.csv`. `visit_log.csv` читается по одной строке, поэтому размер файла не зависит от объёма оперативной памяти.

```python
with open('visit_log.csv', 'r', encoding='utf-8', newline='') as visit_file, \
     open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel_file:

    reader = csv.DictReader(visit_file)
    writer = csv.DictWriter(
        funnel_file,
        fieldnames=['user_id', 'source', 'category']
    )

    writer.writeheader()

    for visit in reader:
        category = purchase_log.get(visit['user_id'])

        if category is not None:
            writer.writerow({
                'user_id': visit['user_id'],
                'source': visit['source'],
                'category': category
            })
```

Проверим первые три строки полученного файла:
