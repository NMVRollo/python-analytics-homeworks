## Подготовка данных и функций

Отдельная функция загружает исходные данные и проверяет, что `age` и `bill` имеют числовой тип. Функции визуализации получают таблицу аргументом; глобальные переменные для хранения данных не используются.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px


def load_clients(csv_path, rows=15):
    clients = pd.read_csv(csv_path, nrows=rows)
    clients['age'] = pd.to_numeric(clients['age'], errors='raise')
    clients['bill'] = pd.to_numeric(clients['bill'], errors='raise')
    return clients


def plot_seaborn_scatter(clients):
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.scatterplot(data=clients, x='age', y='bill', ax=ax)
    ax.set(xlabel='age', ylabel='bill', title='Возраст и расходы клиентов (Seaborn)')
    plt.show()


def plot_plotly_scatter(clients):
    fig = px.scatter(
        clients,
        x='age',
        y='bill',
        labels={'age': 'age', 'bill': 'bill'},
        title='Возраст и расходы клиентов (Plotly)',
    )
    fig.show()


def plot_seaborn_regression(clients):
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.regplot(data=clients, x='age', y='bill', ax=ax)
    ax.set(xlabel='age', ylabel='bill', title='Линия регрессии: возраст и расходы (Seaborn)')
    plt.show()
```

## Задание 1. Диаграмма разброса

Построить диаграмму зависимости возраста клиентов (`age`) от расходов (`bill`) с помощью **Seaborn** и **Plotly**.

### Seaborn: scatterplot

```python
def run_seaborn_scatter(csv_path):
    clients = load_clients(csv_path)
    plot_seaborn_scatter(clients)


run_seaborn_scatter('web_clients_correct.csv')
```

### Plotly: scatterplot (интерактивная диаграмма)

```python
def run_plotly_scatter(csv_path):
    clients = load_clients(csv_path)
    plot_plotly_scatter(clients)


run_plotly_scatter('web_clients_correct.csv')
```

## Задание 2. Линия регрессии

Построить с помощью Seaborn диаграмму `regplot`, которая показывает точки и линию линейной регрессии по тем же первым 15 записям.

```python
def run_seaborn_regression(csv_path):
    clients = load_clients(csv_path)
    plot_seaborn_regression(clients)


run_seaborn_regression('web_clients_correct.csv')
```
