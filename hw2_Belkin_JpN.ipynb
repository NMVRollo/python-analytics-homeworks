## Задание 1. Определение високосного года

Дана переменная, в которой хранится четырёхзначное число - год. 
Нужно определить, является ли этот год високосным или обычным.

```python
year = 2024

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Високосный год")
else:
    print("Обычный год")
```

Проверка второго примера из задания:

```python
year = 2023

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Високосный год")
else:
    print("Обычный год")
```

## Задание 2. Счастливый билет

Дана переменная, в которой хранится шестизначное число - номер проездного билета. 
Билет считается счастливым, если сумма первых трёх цифр совпадает с суммой последних трёх цифр.

```python
n = 123456

number_str = str(n)

first_half = int(number_str[0]) + int(number_str[1]) + int(number_str[2])
second_half = int(number_str[3]) + int(number_str[4]) + int(number_str[5])

if first_half == second_half:
    print("Счастливый билет")
else:
    print("Несчастливый билет")
```

Проверка второго примера из задания:

```python
n = 123321

number_str = str(n)

first_half = int(number_str[0]) + int(number_str[1]) + int(number_str[2])
second_half = int(number_str[3]) + int(number_str[4]) + int(number_str[5])

if first_half == second_half:
    print("Счастливый билет")
else:
    print("Несчастливый билет")
```
