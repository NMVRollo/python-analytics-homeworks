## Задание 1. Средняя буква / две средние буквы

Дана переменная, в которой хранится слово из латинских букв. Если длина слова нечётная, выводится одна средняя буква; если чётная — две средние буквы.


```python
word = 'test'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

Проверка второго примера из задания:


```python
word = 'testing'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

## Задание 2. Идеальные пары

Списки юношей и девушек сортируются по алфавиту. Если количество участников совпадает, формируются пары с одинаковыми индексами. Если количество не совпадает, выводится предупреждение.


```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(f'{boy} и {girl}')
```

Проверка случая, когда количество юношей и девушек не совпадает:


```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(f'{boy} и {girl}')
```
